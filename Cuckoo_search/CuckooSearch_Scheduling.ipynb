```python
import random

# Processing times of 3 jobs on 2 machines
p = [[3, 2],
     [2, 4],
     [4, 1]]

n = len(p)
nests = [random.sample(range(n), n) for _ in range(5)]

# Calculate makespan
def makespan(s):
    m1 = m2 = 0
    for j in s:
        m1 += p[j][0]
        m2 = max(m1, m2) + p[j][1]
    return m2

# Cuckoo Search
best = min(nests, key=makespan)

for _ in range(50):
    for i in range(5):
        new = nests[i][:]
        a, b = random.sample(range(n), 2)
        new[a], new[b] = new[b], new[a]

        j = random.randrange(5)
        if makespan(new) < makespan(nests[j]):
            nests[j] = new

    nests.sort(key=makespan)
    nests[-1] = random.sample(range(n), n)

    if makespan(nests[0]) < makespan(best):
        best = nests[0][:]

print("Best job sequence:", [x+1 for x in best])
print("Minimum makespan:", makespan(best))
```
